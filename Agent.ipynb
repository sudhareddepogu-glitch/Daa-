```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import datetime as dt

from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score

from scipy.cluster.hierarchy import linkage, dendrogram, cut_tree

# ============================================================
# STEP 2: UPLOAD DATASET IN GOOGLE COLAB
# ============================================================

from google.colab import files

uploaded = files.upload()

# Read the uploaded CSV file
retail = pd.read_csv(
    'OnlineRetail.csv',
    sep=',',
    encoding='ISO-8859-1',
    header=0
)

# Display first 5 rows
print("First 5 rows:")
display(retail.head())

# ============================================================
# STEP 3: UNDERSTANDING DATA
# ============================================================

# Shape
print("Shape of dataset:")
print(retail.shape)

# Information
print("\nDataset Information:")
retail.info()

# Description
print("\nDataset Description:")
display(retail.describe())

# ============================================================
# STEP 4: DATA CLEANSING
# ============================================================

# Calculate missing values percentage
df_null = round(100 * retail.isnull().sum() / len(retail), 2)

print("Missing Values Percentage:")
print(df_null)

# Drop rows having missing values
retail = retail.dropna()

print("\nShape after removing missing values:")
print(retail.shape)

# Convert CustomerID to string
retail['CustomerID'] = retail['CustomerID'].astype(str)

# ============================================================
# STEP 5: DATA PREPARATION - RFM ANALYSIS
# ============================================================

# ------------------------------------------------------------
# RFM:
# R = Recency
# F = Frequency
# M = Monetary
# ------------------------------------------------------------

# -------------------------
# Monetary
# -------------------------

retail['Amount'] = retail['Quantity'] * retail['UnitPrice']

rfm_m = retail.groupby('CustomerID')['Amount'].sum()

rfm_m = rfm_m.reset_index()

print("Monetary:")
display(rfm_m.head())

# -------------------------
# Frequency
# -------------------------

rfm_f = retail.groupby('CustomerID')['InvoiceNo'].count()

rfm_f = rfm_f.reset_index()

rfm_f.columns = ['CustomerID', 'Frequency']

print("Frequency:")
display(rfm_f.head())

# -------------------------
# Merge Monetary + Frequency
# -------------------------

rfm = pd.merge(
    rfm_m,
    rfm_f,
    on='CustomerID',
    how='inner'
)

print("RFM after merging Amount and Frequency:")
display(rfm.head())

# ============================================================
# RECENCY
# ============================================================

# Convert InvoiceDate to datetime
retail['InvoiceDate'] = pd.to_datetime(
    retail['InvoiceDate'],
    format='%d-%m-%Y %H:%M'
)

# Maximum/latest date
max_date = retail['InvoiceDate'].max()

print("Maximum Invoice Date:")
print(max_date)

# Difference between maximum date and transaction date
retail['Diff'] = max_date - retail['InvoiceDate']

display(retail.head())

# Minimum difference for each customer
rfm_p = retail.groupby('CustomerID')['Diff'].min()

rfm_p = rfm_p.reset_index()

print("Recency:")
display(rfm_p.head())

# Extract only number of days
rfm_p['Diff'] = rfm_p['Diff'].dt.days

# Merge with RFM dataframe
rfm = pd.merge(
    rfm,
    rfm_p,
    on='CustomerID',
    how='inner'
)

# Rename columns
rfm.columns = [
    'CustomerID',
    'Amount',
    'Frequency',
    'Recency'
]

print("Final RFM Data:")
display(rfm.head())

# ============================================================
# STEP 6: OUTLIER ANALYSIS
# ============================================================

attributes = ['Amount', 'Frequency', 'Recency']

plt.figure(figsize=(10, 8))

sns.boxplot(
    data=rfm[attributes],
    orient='v',
    palette='Set2',
    whis=1.5,
    saturation=1,
    width=0.7
)

plt.title(
    'Outliers Variable Distribution',
    fontsize=14,
    fontweight='bold'
)

plt.ylabel('Range', fontweight='bold')
plt.xlabel('Attributes', fontweight='bold')

plt.show()

# ============================================================
# REMOVE OUTLIERS - AMOUNT
# ============================================================

Q1 = rfm['Amount'].quantile(0.05)
Q3 = rfm['Amount'].quantile(0.95)

IQR = Q3 - Q1

rfm = rfm[
    (rfm['Amount'] >= Q1 - 1.5 * IQR) &
    (rfm['Amount'] <= Q3 + 1.5 * IQR)
]

# ============================================================
# REMOVE OUTLIERS - RECENCY
# ============================================================

Q1 = rfm['Recency'].quantile(0.05)
Q3 = rfm['Recency'].quantile(0.95)

IQR = Q3 - Q1

rfm = rfm[
    (rfm['Recency'] >= Q1 - 1.5 * IQR) &
    (rfm['Recency'] <= Q3 + 1.5 * IQR)
]

# ============================================================
# REMOVE OUTLIERS - FREQUENCY
# ============================================================

Q1 = rfm['Frequency'].quantile(0.05)
Q3 = rfm['Frequency'].quantile(0.95)

IQR = Q3 - Q1

rfm = rfm[
    (rfm['Frequency'] >= Q1 - 1.5 * IQR) &
    (rfm['Frequency'] <= Q3 + 1.5 * IQR)
]

print("Shape after removing outliers:")
print(rfm.shape)

# ============================================================
# STEP 7: RESCALING / STANDARDIZATION
# ============================================================

rfm_df = rfm[['Amount', 'Frequency', 'Recency']]

# Create StandardScaler
scaler = StandardScaler()

# Standardize the data
rfm_df_scaled = scaler.fit_transform(rfm_df)

# Convert back to DataFrame
rfm_df_scaled = pd.DataFrame(rfm_df_scaled)

rfm_df_scaled.columns = [
    'Amount',
    'Frequency',
    'Recency'
]

print("Scaled RFM Data:")
display(rfm_df_scaled.head())

# ============================================================
# STEP 8: K-MEANS CLUSTERING
# ============================================================

# Initial K-Means with K=4
kmeans = KMeans(
    n_clusters=4,
    max_iter=50,
    random_state=42,
    n_init=10
)

kmeans.fit(rfm_df_scaled)

print("K-Means Labels:")
print(kmeans.labels_)

# ============================================================
# STEP 9: ELBOW METHOD
# ============================================================

ssd = []

range_n_clusters = [2, 3, 4, 5, 6, 7, 8]

for num_clusters in range_n_clusters:

    kmeans = KMeans(
        n_clusters=num_clusters,
        max_iter=50,
        random_state=42,
        n_init=10
    )

    kmeans.fit(rfm_df_scaled)

    ssd.append(kmeans.inertia_)

# Plot Elbow Curve
plt.figure(figsize=(8, 5))

plt.plot(
    range_n_clusters,
    ssd,
    marker='o'
)

plt.xlabel('Number of Clusters')
plt.ylabel('SSD / Inertia')
plt.title('Elbow Curve')

plt.show()

# ============================================================
# STEP 10: SILHOUETTE ANALYSIS
# ============================================================

range_n_clusters = [2, 3, 4, 5, 6, 7, 8]

for num_clusters in range_n_clusters:

    kmeans = KMeans(
        n_clusters=num_clusters,
        max_iter=50,
        random_state=42,
        n_init=10
    )

    kmeans.fit(rfm_df_scaled)

    cluster_labels = kmeans.labels_

    silhouette_avg = silhouette_score(
        rfm_df_scaled,
        cluster_labels
    )

    print(
        "For n_clusters = {}, silhouette score = {:.4f}".format(
            num_clusters,
            silhouette_avg
        )
    )

# ============================================================
# STEP 11: FINAL K-MEANS MODEL
# ============================================================

# Use K = 3
kmeans = KMeans(
    n_clusters=3,
    max_iter=50,
    random_state=42,
    n_init=10
)

kmeans.fit(rfm_df_scaled)

# Get cluster labels
labels = kmeans.labels_

print("Final K-Means Labels:")
print(labels)

# Assign labels to original RFM dataframe
rfm['Cluster_Id'] = labels

print("RFM with Cluster ID:")
display(rfm.head())

# ============================================================
# STEP 12: K-MEANS VISUALIZATION
# ============================================================

# Amount vs Cluster
plt.figure(figsize=(8, 5))

sns.boxplot(
    x='Cluster_Id',
    y='Amount',
    data=rfm
)

plt.title('Cluster Id vs Amount')
plt.show()

# Frequency vs Cluster
plt.figure(figsize=(8, 5))

sns.boxplot(
    x='Cluster_Id',
    y='Frequency',
    data=rfm
)

plt.title('Cluster Id vs Frequency')
plt.show()

# Recency vs Cluster
plt.figure(figsize=(8, 5))

sns.boxplot(
    x='Cluster_Id',
    y='Recency',
    data=rfm
)

plt.title('Cluster Id vs Recency')
plt.show()

# ============================================================
# STEP 13: HIERARCHICAL CLUSTERING
# ============================================================

# ------------------------------------------------------------
# SINGLE LINKAGE
# ------------------------------------------------------------

mergings = linkage(
    rfm_df_scaled,
    method='single',
    metric='euclidean'
)

plt.figure(figsize=(12, 6))

dendrogram(mergings)

plt.title('Hierarchical Clustering - Single Linkage')
plt.xlabel('Data Points')
plt.ylabel('Euclidean Distance')

plt.show()

# ------------------------------------------------------------
# COMPLETE LINKAGE
# ------------------------------------------------------------

mergings = linkage(
    rfm_df_scaled,
    method='complete',
    metric='euclidean'
)

plt.figure(figsize=(12, 6))

dendrogram(mergings)

plt.title('Hierarchical Clustering - Complete Linkage')
plt.xlabel('Data Points')
plt.ylabel('Euclidean Distance')

plt.show()

# ------------------------------------------------------------
# AVERAGE LINKAGE
# ------------------------------------------------------------

mergings = linkage(
    rfm_df_scaled,
    method='average',
    metric='euclidean'
)

plt.figure(figsize=(12, 6))

dendrogram(mergings)

plt.title('Hierarchical Clustering - Average Linkage')
plt.xlabel('Data Points')
plt.ylabel('Euclidean Distance')

plt.show()

# ============================================================
# STEP 14: CUT DENDROGRAM INTO 3 CLUSTERS
# ============================================================

cluster_labels = cut_tree(
    mergings,
    n_clusters=3
).reshape(-1)

print("Hierarchical Cluster Labels:")
print(cluster_labels)

# Assign cluster labels
rfm['Cluster_Labels'] = cluster_labels

print("RFM with Hierarchical Cluster Labels:")
display(rfm.head())

# ============================================================
# STEP 15: HIERARCHICAL CLUSTER VISUALIZATION
# ============================================================

# Amount
plt.figure(figsize=(8, 5))

sns.boxplot(
    x='Cluster_Labels',
    y='Amount',
    data=rfm
)

plt.title('Hierarchical Cluster vs Amount')
plt.show()

# Frequency
plt.figure(figsize=(8, 5))

sns.boxplot(
    x='Cluster_Labels',
    y='Frequency',
    data=rfm
)

plt.title('Hierarchical Cluster vs Frequency')
plt.show()

# Recency
plt.figure(figsize=(8, 5))

sns.boxplot(
    x='Cluster_Labels',
    y='Recency',
    data=rfm
)

plt.title('Hierarchical Cluster vs Recency')
plt.show()

# ============================================================
# STEP 16: CLUSTER SUMMARY
# ============================================================

print("K-Means Cluster Summary:")

kmeans_summary = rfm.groupby('Cluster_Id')[
    ['Amount', 'Frequency', 'Recency']
].mean()

display(kmeans_summary)

print("Hierarchical Cluster Summary:")

hierarchical_summary = rfm.groupby('Cluster_Labels')[
    ['Amount', 'Frequency', 'Recency']
].mean()

display(hierarchical_summary)
```
